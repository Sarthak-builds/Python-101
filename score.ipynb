```python
from langgraph.graph import StateGraph, START, END
from IPython.display import Image
from typing import TypedDict
```

```python
# a single batting score state
class ScoreState(TypedDict):
    runs: int
    balls: int
    sixes: int
    fours: int
    summary: str
```

```python
def compute_summary(state: ScoreState) -> dict:
    runs = state['runs']
    balls = state['balls']
    fours = state['fours']
    sixes = state['sixes']

    sr = round((runs / balls) * 100, 2) if balls > 0 else 0.0
    boundary_runs = fours * 4 + sixes * 6
    boundary_percent = round((boundary_runs / runs) * 100, 2) if runs > 0 else 0.0

    summary = (
        f"Runs: {runs} | Balls: {balls} | 4s: {fours} | 6s: {sixes}\n"
        f"Strike rate: {sr} | Boundary percent: {boundary_percent}%"
    )
    return {'summary': summary}
```

```python
graph = StateGraph(ScoreState)

graph.add_node('compute_summary', compute_summary)

graph.add_edge(START, 'compute_summary')
graph.add_edge('compute_summary', END)

workflow = graph.compile()
```

```python
workflow.invoke({'runs': 87, 'balls': 52, 'fours': 7, 'sixes': 3})
```
